```python
def hitung_pendapatan(berat_kg, harga_per_kg):
    """Menghitung total pendapatan kotor sebelum diskon."""
    return berat_kg * harga_per_kg


def hitung_diskon(total_pendapatan, persentase_diskon):
    """Menghitung nominal potongan harga berdasarkan persentase."""
    return total_pendapatan * (persentase_diskon / 100)


def hitung_total_setelah_diskon(total_pendapatan, nominal_diskon):
    """Menghitung total pendapatan bersih yang diterima setelah dipotong diskon."""
    return total_pendapatan - nominal_diskon


if __name__ == "__main__":
    komoditas = "Kentang"
    berat_panen = 1200       # dalam kilogram (kg)
    harga_per_kg = 14000     # harga per kg (Rp)
    persen_diskon = 25        # diskon 25% (misal untuk pembeli grosir/tengkulak)

    # 1. Hitung total pendapatan kotor
    pendapatan_kotor = hitung_pendapatan(berat_panen, harga_per_kg)

    # 2. Hitung nominal diskon dari pendapatan kotor tersebut
    nominal_diskon = hitung_diskon(pendapatan_kotor, persen_diskon)

    # 3. Hitung pendapatan akhir yang diterima
    pendapatan_bersih = hitung_total_setelah_diskon(pendapatan_kotor, nominal_diskon)

    # Tampilkan rincian laporan
    print("=" * 45)
    print(f"   LAPORAN PENDAPATAN PANEN {komoditas.upper()}")
    print("=" * 45)
    print(f"Komoditas            : {komoditas}")
    print(f"Total Hasil Panen    : {berat_panen:,} kg")
    print(f"Harga Satuan         : Rp {harga_per_kg:,} / kg")
    print("-" * 45)
    print(f"Total Pendapatan     : Rp {pendapatan_kotor:,}")
    print(f"Diskon ({persen_diskon}%)          : -Rp {nominal_diskon:,.0f}")
    print("-" * 45)
    print(f"Total yang Diterima  : Rp {pendapatan_bersih:,.0f}")
    print("=" * 45)
```
